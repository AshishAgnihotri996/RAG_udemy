## what are embeddigs

```python
%pip install matplotlib

import numpy as np 
import matplotlib.pyplot as plt
```

```python
word_embeddings = {
    "cat": [0.1, 0.2, 0.3],
    "kitten": [0.1, 0.2, 0.25],
    "dog": [0.4, 0.5, 0.6],
    "cow": [0.5, 0.6, 0.7],
    "car": [0.8, 0.9, 1.0],
    "truck": [1.1, 1.2, 1.3]
}
```

```python
fix, ax = plt.subplots(figsize=(8, 6))
for word,coords in word_embeddings.items():
    ax.scatter(*coords, label=word)
    ax.annotate(word, (coords[0], coords[1]), textcoords="offset points", xytext=(10, 10), ha='center')

ax.set_xlabel('Dimension 1')    
ax.set_ylabel('Dimension 2')    
ax.set_title('Word Embeddings Visualization')
ax.legend()
ax.grid(True,alpha=0.3)

plt.tight_layout
plt.show()  
```
