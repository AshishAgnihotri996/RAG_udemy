```python
os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")
```

## openai embeddings

```python
from langchain_openai import OpenAIEmbeddings
embeddings = OpenAIEmbeddings(model="minimax/minimax-m3.1-flash:free")
##SET THE OPENAI_API_KEY in the .env file and then run this code to get the embeddings for the text.
```

```python
embeddings
```

```python
##single text embedding
single_text = "This is a sample text for embedding."
single_embedding = embeddings.embed_query(single_text)
print(len(single_embedding))
print(single_embedding)
```
